```python
import os
from typing import Optional
import shutil
from datetime import datetime
from dotenv import load_dotenv
from supabase import create_client, Client

load_dotenv(".//.env")

#Upload to blob directly
license_number = "MH-12-UY-0916" 
supabase: Client = create_client(os.getenv("SUPABASE_URL"),os.getenv("SUPABASE_SECRET_KEY"))
folder_path_raw = r"C:\Users\Samuel\Desktop\SAM\CODING\PYTHON\LicensePlateRecognition\License-Plate-Detector\backend\scans\2026-10-13_19-43-21"
if folder_path_raw =="" or license_number  =="":
    print("Error fetching status")
date = datetime.now().strftime("%d-%m-%Y")
time = datetime.now().strftime("%H-%M-%S")
vehicle_time =f'capture_log/{date}/{license_number}_{time}'
bucket_name = os.getenv("S3_BUCKET")
supabase_url = os.getenv("SUPABASE_URL")
for filename in ["Captured_Image.jpg", "Car_Crop_Capture.jpg"]:
    file = folder_path_raw +"/"+ filename
    output_file = vehicle_time +"/"+ filename
    with open(file, "rb") as f:
        supabase.storage.from_(bucket_name).upload(file=f,
        path=output_file,  
        file_options={"cache-control": "3600", "upsert": "false", "content-type":"image/jpeg"})
    #cleanup
blobURL = supabase_url + "/storage/v1/object/public/" + bucket_name+"/"+vehicle_time
print(blobURL)
mypath ="https://supabase.com/dashboard/project/smxmlxxaifxqckxnaqce/storage/files/buckets/parking-logs?path="+vehicle_time
print("Direct access: ",mypath)
```
